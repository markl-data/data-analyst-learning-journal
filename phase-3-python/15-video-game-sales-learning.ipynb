## Q6 Findings: Top 10 Games Contribution

**The industry is franchise-driven, not hit-driven.**

The top 10 best-selling games globally represent only 4.14% of total 
industry sales ($369.39M of $8,920.44M total). This is dramatically 
lower than either an anticipated hit-driven (15-25%) or moderate 
concentration (10-15%) — the distribution is skewed toward the long tail 
of ~16,000 individual titles.

**Nintendo occupies every top-10 slot.**

All 10 of the highest-selling games globally are Nintendo titles, from 
Wii Sports at $82.74M to Duck Hunt at $28.31M. No other publisher 
appears — despite Nintendo holding only ~20% of overall global market 
share. This is a level of blockbuster dominance no other publisher matches.

**One outlier worth noting:**

Wii Sports at $82.74M is a distinct outlier — 2.05× the next-highest game 
(Super Mario Bros. at $40.24M). Games ranked 2-10 cluster in the 
$28-40M range, showing a flat plateau after the single dominant title.

**Combined insight from Q1 + Q6:**

Q1 established that the top 10 publishers control ~70% of global sales. 
Q6 reveals the top 10 games represent only 4.14%. Together these produce 
a sharp structural insight:

> **The video game market is concentrated at the publisher level but 
> distributed at the game level. Publishers succeed through deep 
> catalogues of many-titles-each, not through reliance on rare blockbusters.**

This has implications for portfolio strategy — publisher success in 
gaming appears tied to catalogue depth and franchise recurrence, not 
individual title breakouts.

**Domain hypotheses (not proven from data):**

Nintendo's dominance in top-10 may reflect platform strategy (Nintendo 
publishes exclusively on Nintendo hardware, capturing the entire 
Nintendo audience), evergreen franchise IP (Mario, Pokemon, Zelda across 
decades), or bundling strategies (Wii Sports came bundled with Wii 
console purchases). The data shows the pattern; the causes require 
different data to test.
