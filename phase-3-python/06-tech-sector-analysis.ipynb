## Technology Sector Deep-Dive — Revised Findings

**Revenue growth patterns (2009-2022):**

Absolute revenue:
- AAPL: $42.9B → $394.3B (9.2× multiple)
- GOOG: $23.7B → $282.8B (12× multiple — highest multiple)
- MSFT: $58.4B → $198.3B (3.4× multiple — lowest multiple)

Growth consistency (YoY volatility):
- AAPL: highly variable (-8% to +66%) — product-cycle-driven
- GOOG: remarkably consistent (+10% to +41%, mostly mid-teens/twenties) — 
  the steadiest grower
- MSFT: moderately variable (+3% to +18%) — steady but modest

**Sector tailwind confirmed** for aggregate years: 2011, 2018, 2020, 2021 
show all three growing together. But individual years diverge:
- 2015-2016: AAPL super-cycle (+28%, then -8%), others modest
- 2017: GOOG accelerated (+23%) while AAPL/MSFT plodded (+6% each)
- 2019: AAPL declined (-2%) while GOOG/MSFT grew (+18%, +14%)

**Net Profit Margin patterns — more nuanced than typical framing:**

| Company   | Range     | Character                                                             |
|---        |---        |---                                                                    |
| MSFT      | 13% - 37% | Highest peaks but wide swings (2015, 2018 dips)                       |
| GOOG      | 11% - 30% | Moderately high; 2017 EU antitrust fine dropped margin to 11%         |
| AAPL      | 19% - 27% | Steadiest — hardware business drag but consistent                     |

**Revised conclusion:** The Tech trio confirms sector-level revenue tailwinds 
in most years, but the "MSFT highest, GOOG middle, AAPL lowest" margin story 
is too simple. AAPL is actually the most stable margin performer; MSFT has 
higher peaks but more volatility. Each company reflects distinct business 
model dynamics within a shared sector environment.
